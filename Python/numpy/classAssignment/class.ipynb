# 1. Declaring Arrays

````python
from nptyping import NDArray
import numpy as np
# from typing import Any

a : NDArray = np.array(1000)
a1D: NDArray = np.array([1, 2, 3, 4])
a2D: NDArray = np.array([[1, 2], [3, 4]])
a3D: NDArray = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print(a)
print(a1D)
print(a2D)
print(a3D)
````

# 2. np.inid()

````python
def elements_in_array(arr1: NDArray, arr2: NDArray) -> NDArray:
    return np.in1d(arr1, arr2)

print(elements_in_array([1,2,4], [1,5,4]))
````

# 3. np.intersect1d()

````python
def intersection_of_arrays(arr1: NDArray, arr2: NDArray) -> NDArray:
    return np.intersect1d(arr1, arr2)

print(intersection_of_arrays([1,2,4], [1,5,4]))
````

# 4. np.arange()

````python
x: NDArray = np.arange(10)
print(x)
print(x[-3])
print(x[3])
````

# 5. np.cos()

````python
x: NDArray = np.cos([0,np.pi,np.pi/4,np.pi/2])
print(x)
print(x[1])
print(x[-1])
````

# 6. np.pi

````python
a: float = np.pi
a
````

# 7. np.insert()

````python
x = np.insert([0,1,2], 1,[5,6,7,8])
x
````

# 8. np.array_str

````python
b:str = np.array_str(x)
b
````

# 9. np.transpose

````python
a: NDArray = [[3,4,5], [1,2,3]]
np.transpose(a)
````

# 10. np.split

````python
x = np.arange(12)
np.split(x,6)
````

# 11. np.repeat

````python
print(np.repeat(3, 4))
x: NDArray = [[1,2],[3,4]]
print(np.repeat(x, 2))
x: NDArray = [[1,2,3,4,5,6]]
print(np.repeat(x, 2))
````
